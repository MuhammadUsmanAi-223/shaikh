# Lab 1: Data Preprocessing
## Task 1 - Dataset Cleaning
- Load the Titanic dataset (`train.csv`)
- Display the first 10 rows
- Check for missing values in each column
- Fill missing values in the "Age" column with the mean age
- Drop rows where the "Embarked" column is missing

```python
import pandas as pd

# 1. Load the Titanic dataset (train.csv)
df = pd.read_csv('Train.csv')

# 2. Display the first 10 rows
print("--- First 10 Rows ---")
print(df.head(10))

# 3. Check for missing values in each column
print("\n--- Missing Values Before Cleaning ---")
print(df.isnull().sum())

# 4. Fill missing values in the "Age" column with the mean age
df['Age'] = df['Age'].fillna(df['Age'].mean())

# 5. Drop rows where the "Embarked" column is missing
df = df.dropna(subset=['Embarked'])

# Verify missing values after cleaning
print("\n--- Missing Values After Cleaning ---")
print(df.isnull().sum())
```

## Task 2 - Encoding Categorical Data
- Convert the "Sex" column into numeric (0 = Male, 1 = Female)
- Apply One-Hot Encoding on the "Embarked" column

```python
# Convert the "Sex" column into numeric (0 = Male, 1 = Female)
df['Sex'] = df['Sex'].map({'male': 0, 'female': 1})

# Apply One-Hot Encoding on the "Embarked" column
df = pd.get_dummies(df, columns=['Embarked'], drop_first=False)

# Display updated dataset columns
print("--- Encoded Features Sample ---")
print(df[['Sex', 'Embarked_C', 'Embarked_Q', 'Embarked_S']].head())
```

## Task 3 - Feature Scaling & Splitting
- Select features: Age, Fare, Sex, Pclass
- Apply StandardScaler to normalize them
# - Split data into 80% training and 20% testing sets

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Select features: Age, Fare, Sex, Pclass
X = df[['Age', 'Fare', 'Sex', 'Pclass']]
y = df['Survived']

# Apply StandardScaler to normalize numerical features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Convert scaled arrays back into a DataFrame for visualization
X_scaled_df = pd.DataFrame(X_scaled, columns=X.columns)

# Split data into 80% training and 20% testing
X_train, X_test, y_train, y_test = train_test_split(
    X_scaled_df, y, test_size=0.20, random_state=42
)

# Output dataset shapes
print(f"X_train shape: {X_train.shape}")
print(f"X_test shape:  {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape:  {y_test.shape}")
```
